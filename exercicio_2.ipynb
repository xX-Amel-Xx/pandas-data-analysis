```python
import pandas as pd
import numpy as np

df=pd.read_csv("funcionarios.csv")
display(df)
```

```python
#Exercício 1
# Mostre apenas os funcionários que têm idade acima da média.

media_idade=df["Idade"].mean()
analise_funcionarios=df[df["Idade"]>media_idade]
display(analise_funcionarios)
```

```python
#Exercício 2
# Mostre os funcionários que ganham entre 3500 e 5000.
analise_funcionarios=df[(df["Salario"]>=3500) &(df["Salario"]<=5000)]
display(analise_funcionarios)
```

```python
#Exercício 3
# Mostre os funcionários que trabalham em TI ou RH.
analise_funcionarios=df[(df["Departamento"]=="TI") | (df["Departamento"]=="RH")]
display(analise_funcionarios)
```

```python
#Exercício 4
# Mostre apenas os funcionários cuja cidade não é Brasília.
analise_funcionarios=df[df["Cidade"]!="Brasília"]
display(analise_funcionarios)
```

```python
#Exercício 5
#Mostre apenas as colunas:Nome, Cidade, dos funcionários com salário acima de 4500.

analise_funcionarios=df[df["Salario"]>4500] [["Nome","Cidade"]]
display(analise_funcionarios)
```

```python
#Exercício 6
#Crie uma coluna chamada Bonus.
#O bônus será 10% do salário.
df["Bonus"]=df["Salario"]*0.1
display(df)
```

```python
#Exercício 7
#Crie uma coluna chamada Salario_Final.
df["Salario_Final"]=df["Salario"]+df["Bonus"]
display(df)
```

```python
#Exercício 8
#Mostre apenas os 5 maiores salários.
analise_salario=df.sort_values(by="Salario",ascending=False)
display(analise_salario.head(5))
```

```python
#Exercício 9
#Mostre apenas os 5 menores salários.
analise_salario=df.nsmallest(5,"Salario")
display(analise_salario)
```

```python
#Exercício 10
#Calcule para cada departamento:
#Média salarial, Maior salário, Menor salário, (Dica: use agg().

analise_salario=df.groupby("Departamento").agg(Media=("Salario","mean"),
                                                           Maior_salario=("Salario","max"),
                                                           Menor_salario=("Salario","min"))
display(analise_funcionarios)
```

```python
#Exercício 11
#Qual cidade possui a maior média salarial?
media_cidade=df.groupby("Cidade")["Salario"].mean()
analise_cidade=media_cidade.idxmax()
display(analise_cidade)
```

```python
#Exercício 12
# Mostre a soma dos salários de cada departamento.
analise_salario=df.groupby("Departamento") ["Salario"].sum()
display(analise_salario)
```

```python
#Exercício 13
# Ordene os departamentos pela soma dos salários (maior para menor).
analise_salario=df.groupby("Departamento")["Salario"].sum()
analise_salario=analise_salario.sort_values(ascending=False)
display(analise_salario)
```

```python
#Exercício 14
# Mostre os funcionários cujo nome começa com "M".
analise_funcionarios=df[df["Nome"].str.startswith("M")]
display(analise_funcionarios)
```

```python
#Exercício 15
# Mostre os funcionários cujo nome termina com "a".
analise_funcionarios=df[df["Nome"].str.endswith("a")]
display(analise_funcionarios)
```

```python
#Exercício 16
# Mostre todos os funcionários em ordem alfabética.
analise_funcionarios=df.sort_values(by="Nome",ascending=True)
display(analise_funcionarios)
```

```python
#Exercício 17
#Mostre os funcionários que possuem idade entre 25 e 35 anos.
analise_funcionarios=df[(df["Idade"]>=25) & (df["Idade"]<=35)]
display(analise_funcionarios)
```

```python
#Exercício 18
# Crie uma coluna chamada Faixa_Etaria:
# Até 25 → Jovem
# 26 até 35 → Adulto
# Acima de 35 → Experiente

df["Faixa_Etaria"]=np.where(df["Idade"]<=25,"Jovem", 
np.where(df["Idade"].between(26,35),"Adulto","Experiente"))
display(df)
```

```python
#Exercício 19
# Qual departamento possui o maior salário médio?
media_salario=df.groupby("Departamento")["Salario"].mean()
analise_salario=media_salario.idxmax()
display(analise_salario)
```

```python
#Desafio 1 Mostre:
# Cidade
# Quantidade de funcionários
# Média salarial
# Tudo em uma única tabela.

df_analise=df.groupby("Cidade").agg(
    quantidade_funcionarios=("Nome","count"),media_salarial=("Salario","mean")) 
display(df_analise)
```

```python
#Desafio 2 
# Para cada departamento, mostre:
# Quantidade de funcionários
# Média salarial
# Maior salário
# Menor salário
# Soma dos salários
# Tudo usando apenas um groupby().

df_analise=df.groupby("Departamento").agg(QUantidade_Funcionario=("Nome","count"),Media_salario=("Salario","mean"), Maior_Salario=("Salario","max"),
                                         Menor_Salario=("Salario","min"), Soma_Salario=("Salario","sum"))
display(df_analise)
```

```python
#Desafio 3
# Crie um ranking do maior para o menor salário e exiba:
# Nome,Departamento,Salário,Ranking
df["Ranking"]=df["Salario"].rank(ascending=False, method="dense")

df_analise=df[["Nome","Departamento","Salario","Ranking"]]
display(df_analise)
```
