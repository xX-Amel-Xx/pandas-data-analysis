```python
import pandas as pd
df=pd.read_csv("funcionarios.csv")
display(df)
```

```python
#1. sum()
# Exercício:
# Calcule o total de salários pagos por cada departamento.
analise_salario=df.groupby("Departamento")["Salario"].sum()
display(analise_salario)
```

```python
#2. mean()
#Exercício:
# Calcule a média salarial de cada cidade.
# Dica: groupby() + mean().
analise_salario=df.groupby("Cidade")["Salario"].mean()
display(analise_salario)
```

```python
#3. count()
# Exercício:
# Conte quantos funcionários existem em cada departamento.
# Dica: você pode contar a coluna Nome.
analise_funcionario=df.groupby("Departamento")["Nome"].count()
display(analise_funcionario)
```

```python
#4. max()
#Exercício:
#Mostre o maior salário de cada departamento.
analise_salario=df.groupby("Departamento")["Salario"].max()
display(analise_salario)
```

```python
#5. min()
#Exercício:
#Mostre o menor salário de cada departamento.
analise_salario=df.groupby("Departamento")["Salario"].min()
display(analise_salario)
```

```python
#6. agg()
# Exercício:
#Para cada departamento, mostre em uma única tabela:
#Média salarial
#Maior salário
#Menor salário
analise_salario=df.groupby("Departamento").agg(Media_salarial=("Salario","mean"),
                                               Maior_saiario=("Salario","max"),
                                               menor_salario=("Salario","min")
                                               )
display(analise_salario)
```

```python
#7. Agrupar por duas colunas
#Exercício:
#Descubra a média salarial por cidade e departamento.
#O resultado deverá ter uma estrutura parecida com:

analise_salario=df.groupby(["Cidade","Departamento"])["Salario"].mean()
display(analise_salario)
```
