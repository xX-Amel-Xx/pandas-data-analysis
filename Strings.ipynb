```python
import pandas as pd
df=pd.read_csv("funcionarios.csv")
display(df)
```

```python
#1. Mostre todos os funcionários cujo Nome contém a letra "a"
#2. Mostre os funcionários cujo Nome contém "an".
#3. Crie uma coluna chamada Tem_A indicando True ou False para 
# cada funcionário, dependendo se o nome contém a letra "a".
a=df[df["Nome"].str.contains("a")]
an=df[df["Nome"].str.contains("an")]
df["Tem_A"]=df["Nome"].str.contains("a")
display(df)
display(a,an)
```

```python
# lower()
#4. Crie uma coluna chamada Nome_Minusculo contendo todos os nomes em letras minúsculas.
#5. Crie uma coluna Departamento_Minusculo com os departamentos em letras minúsculas.

df["Nome_Maiúsculo"]=df["Nome"].str.upper()
df["Nome_Minúsculo"]=df["Nome"].str.lower()
display(df)
```

```python
#Para esse exercício, primeiro crie uma pequena coluna com espaços extras.
#8. Crie:
#" Amanda "," João","Maria "," Pedro "
#Depois remova os espaços do começo e do final usando strip().
df_nomes = pd.DataFrame({
    "Nomes": [" Amanda ", " João", "Maria ", " Pedro "]
})
df_nomes["Nomes"]=df_nomes["Nomes"].str.strip()
display(df_nomes)
```

```python
#replace()
#9. Na coluna Departamento, troque:
#"TI" → "Tecnologia"
#10. Troque:
#"RH" → "Recursos Humanos"
df["Departamento_Novo"]=df["Departamento"].str.replace("TI","Tecnologia")
df["Departamento_Novo"]=df["Departamento_Novo"].str.replace("RH","Recursos Humanos")
display(df)
```

```python
#split()
#12. Pegue a coluna Nome e transforme cada nome em uma lista de palavras.
nomes=pd.DataFrame({
    "Nome":["Amanda Meireles","Stuart Raw","Lila Olivia"]
})

nomes["Nome"]=nomes["Nome"].str.split(" ")
display(nomes)
```
