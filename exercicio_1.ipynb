```python
# Leia o arquivo CSV.

import pandas as pd
import numpy as np

df=pd.read_csv("funcionarios.csv")
display(df)
```

```python
#Mostre: As 5 primeiras linhas. As 5 últimas linhas.

display(df.head(5))
display(df.tail(5))
```

```python
# Quantas linhas existem.
# Quantas colunas existem.
# Os tipos de dados de cada coluna.

display(df.shape[0])
display(df.shape[1])
display(df.info)
```

```python
#Mostre apenas as colunas:
#Nome
#Salario
df=df[["Nome","Salario"]]
display(df)
```

```python
#Mostre apenas os funcionários que: Ganham mais de 4000.

df_salario=df[df["Salario"]>4000]
display(df_salario)
```

```python
df_regiao=df[df["Cidade"]=="Brasília"]
display(df_regiao)
```

```python
#Mostre quem: Mora em Goiânia e trabalha em TI.

analise_regiao=df[(df["Cidade"]=="Rio de Janeiro") & (df["Departamento"]=="TI")]
display(analise_regiao)
```

```python
#Ordene os salários: Do menor para o maior.Do maior para o menor.

maior=df.sort_values(by="Salario", ascending=False)
menor=df.sort_values(by="Salario", ascending=True)
display(maior)
display(menor)
```

```python
#Crie uma coluna chamada Salario_Anual.
df["salario_anual"]=df["Salario"]*12
display(df)
```

```python
#Crie uma coluna chamada Categoria:
#Salário < 3000 → "Baixo"
#Salário entre 3000 e 5000 → "Médio"
#Salário ≥ 5000 → "Alto"
df["Categorias"]=np.where(
    df["Salario"] <3000, 
    "Baixo", 
np.where(
    (df["Salario"]>= 3000) & (df["Salario"] <5000),
"Médio", "Alto"))
display(df)
```

```python
#Descubra:
#Média dos salários.
#Maior salário.
#Menor salário.

df["Média Salário"]=df["Salario"].mean()
df["Maior Salário"]=df["Salario"].max()
df["Menor Salário"]=df["Salario"].min()

display(df)
```

```python
#Mostre apenas os funcionários com salário acima da média.
media=df["Salario"].mean()
Fmedia=df[df["Salario"]>media]
display(Fmedia)
```

```python
#Conte quantos funcionários existem em cada cidade.
fun=df.groupby("Cidade")["Nome"].count()
display(fun)
```

```python
#Conte quantos funcionários existem em cada departamento.
qtd_fun=df.groupby("Departamento")["Nome"].count()
display(qtd_fun)
```

```python
#Mostre apenas: Nome, Departamento, Salario dos funcionários que trabalham em Vendas.

df_analise=df[df["Departamento"].isin(["Vendas"])][["Nome","Departamento","Salario"]]
display(df_analise)
```

```python
#Mostre o funcionário com o maior salário.
df_analise=df[df["Salario"]==df["Salario"].max()]["Nome"]
display(df_analise)
```

```python
#Mostre os 3 funcionários com os maiores salários.
df_analise=df.nlargest(3,"Salario")["Nome"]
display(df_analise)
```

```python
#Calcule a média salarial de cada departamento usando groupby().
media_slr=df.groupby("Departamento")["Salario"].mean()
display(media_slr)
```
