```python
import pandas as pd
import numpy as np


dados = {
    "Nome": ["Amanda", "João", "Maria", "Pedro", "Ana"],
    "Idade": [21, 30, 25, 40, 28],
    "Cidade": ["Brasília", "São Paulo", "Rio de Janeiro", "Brasília", "Goiânia"],
    "Salario": [2500, 4200, 3800, 5000, 3100]
}

df = pd.DataFrame(dados)

display(df)
```

```python
display(df.head(2))
display(df.head(3))
```

```python
display(df[["Nome", "Salario"]])
nova=df.sort_values(by=["Salario"],ascending=True)
display(nova)
```

```python
display(df.loc[0])
display(df.loc[2])
display(df.loc[1:4])
```

```python
salario_maior=df[df["Salario"]>3000]
display(salario_maior["Nome"].unique())

display(salario_maior)
Regiao=df[df["Cidade"]=="Brasília"]
display(Regiao)

idade=df[df["Idade"]<30]
display(idade)
```

```python
Regiao_Salario=df[(df["Cidade"]=="Brasília") & (df["Salario"]>3000)]
display(Regiao_Salario)
```

```python
analise={
"Media dos salarios":df["Salario"].mean(),
"Maior valor":df["Salario"].max(),
"Menor valor":df["Salario"].min(),
"Média das Idades":df["Idade"].mean()
}
display(analise)
```

```python
dic_Salario={
"Maior Salário": df.sort_values(by=["Salario"],ascending=False),
"Menor Salário":df.sort_values(by=["Salario"],ascending=True),
"Salario Idade":df.sort_values(by=["Idade"], ascending=False)
}
display(dic_Salario)
```

```python
df["Salario_Anual"]=df["Salario"]*12
display(df)
```

```python
df=df.drop(columns=["Cidade"])
display(df)
```

```python
media=df["Salario"].mean()

Acima_Media=df[df["Salario"]>media]
display(Acima_Media)
```

```python
df["Categoria"]=np.where(
    df["Salario"]<3000, 
    "Baixo",
np.where(
    (df["Salario"]>=3000) & (df["Salario"]<4000), 
    "Médio",
    "Alto"))

display(df)
analise_salario=df.sort_values(by="Salario",ascending=False)[["Nome","Salario","Categoria"]]
display(analise_salario)
```
