```python
import pandas as pd
fun=pd.read_csv("funcionarios_datas.csv")
display(fun)

fun["Data"]=pd.to_datetime(fun["Data"],dayfirst=True)
display(fun["Data"].dtype)
```

```python
#5. Crie uma coluna chamada Ano contendo somente o ano da Data.
fun["Ano"]=fun["Data"].dt.year
display(fun)
```

```python
#6. Crie uma coluna chamada Mes contendo o número do mês.
fun["Mês"]=fun["Data"].dt.month
display(fun)
```

```python
#7. Crie uma coluna chamada Dia contendo o dia.
fun["Dia"]=fun["Data"].dt.day
display(fun)
```

```python
#8. Crie uma coluna chamada Dia_Semana contendo o nome do dia da semana.
fun["Dia da Semana"]=fun["Data"].dt.day_name()
display(fun)
```

```python
janeiro=fun[fun["Data"].dt.month==1]
display(fun)
```

```python
#10. Mostre somente os funcionários registrados no mês de maio.
maio=fun[fun["Data"].dt.month==5]
display(maio)
```

```python
#11. Mostre somente os funcionários registrados no ano de 2026.
ano_2026=fun[fun["Data"].dt.year==2026]
display(ano_2026)
```

```python
#12. Mostre somente os registros que aconteceram em uma segunda-feira.
segunda=fun[fun["Data"].dt.day_name()=="Monday"]
display(segunda)
```

```python
#13. Conte quantos funcionários foram registrados em cada mês.
cont_fun=fun["Data"].dt.month.value_counts()
display(cont_fun)
```

```python
#14. Conte quantos funcionários foram registrados em cada dia da semana.
cont_fun=fun["Data"].dt.day_of_week.value_counts()
display(cont_fun)
```

```python
#15. Conte quantos funcionários foram registrados em cada departamento por mês.
fun["Mes"]=fun["Data"].dt.month
cont_fun=fun.groupby(["Departamento","Mes"]).size()
display(cont_fun)
```

```python
#17. Descubra qual mês teve mais registros.
#18. Descubra qual foi a data mais antiga da tabela.
#19. Descubra qual foi a data mais recente.
#20. Ordene os funcionários da data mais antiga para a mais recente.

fun["Mes_ano"]=fun["Data"].dt.strftime("%m/%Y")
resultado=fun["Mes_ano"].value_counts().idxmax()
display(resultado)

display(fun["Mes_ano"].min())

display(fun["Mes_ano"].max())


display(fun.sort_values(by="Mes_ano", ascending=True))
```
