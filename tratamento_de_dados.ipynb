```python
import pandas as pd
import numpy as np

df=pd.read_csv("funcionarios.csv")
display(df)
```

```python
#Exercício 1 — Explorando os dados
#Leia o arquivo CSV e responda:
#Quantas linhas existem?
#Quantas colunas existem?
#Quais são os tipos de dados?
#Existem valores nulos?

display(df.shape[0])
display(df.shape[1])

display(df.info())
display(df.isnull().sum())

```

```python
#Exercício 2 — Valores nulos
# Mostre apenas as linhas que possuem pelo menos um valor nulo.

df_analise=df[df.isnull().any(axis=1)]
display(df_analise)
```

```python
#Exercício 3 — Valores preenchidos
#Mostre apenas as linhas que não possuem nenhum valor nulo.
#Objetivo: notnull().
df_analise=df[df.notnull().all(axis=1)]
display(df_analise)
```

```python
#Exercício 4 — Contando nulos
# Descubra:Quantos valores nulos existem em cada coluna.
df_analise=df.isnull().sum()
display(df_analise)
```

```python
#Exercício 5 — Preenchendo valores
# Preencha:
# Salários nulos com 0.
# Cidade nula com "Não informado".
# Objetivo: fillna().

df["Salario"]=df["Salario"].fillna(0)
df["Cidade"]=df["Cidade"].fillna("Não informado")
display(df)
```

```python
#Exercício 6 — Removendo linhas
#Remova todas as linhas que possuem algum valor nulo.
#Objetivo: dropna().
display(df.dropna())
```

```python
#Exercício 7 — Removendo apenas quando faltar salário
#Remova somente as linhas onde o Salario é nulo.
analise_salario=df.dropna(subset=["Salario"])
display(analise_salario)

```

```python
#Exercício 8 — Valores duplicados 
# Descubra: Existem funcionários duplicados?
df_analise=df["Nome"].duplicated().any()
display(df_analise)

```

```python
#Exercício 9 — Remover duplicados
#Remova todos os registros duplicados.
df=df.drop_duplicates()
display(df)
```

```python
#Exercício 10 — Substituindo valores
# Troque:
# "TI" por "Tecnologia".
df["Departamento"]=df["Departamento"].replace({
"TI":"Tecnologia"
})
display(df)
```

```python
#Exercício 11 — Outra substituição
# Troque:
# "RH" por "Recursos Humanos".
df["Departamento"]=df["Departamento"].replace({
    "RH":"Recursos Humanos"
})
display(df)
```

```python
#Exercício 12 — Alterando tipo
# Transforme a coluna Idade para inteiro
#Objetivo: astype().
df["Idade"]=df["Idade"].astype(int)
display(df.info())
```

```python
#Exercício 13 — Outra conversão
#Transforme a coluna Salario para float.
df["Salario"]=df["Salario"].astype(float)
display(df.dtypes)
```

```python
#Exercício 14 — Tudo junto
# Faça um tratamento completo:
# Remova duplicados.
# Preencha cidades vazias.
# Remova salários nulos.
# Troque "TI" por "Tecnologia".
# Converta salário para float.
df_analise=df.drop_duplicates()
df_analise["Cidade"]=df_analise["Cidade"].fillna("Nao encontrado")
df_analise=df_analise.dropna(subset=["Salario"])
df_analise["Departamento"]=df_analise["Departamento"].replace({
    "Tecnologia":"TI"
})
df_analise["Salario"]=df_analise["Salario"].astype(float)
display(df_analise)
```

```python
#Desafio 1
# Mostre apenas os funcionários que possuem cidade preenchida.
analise_funcionario=df[df["Cidade"].notnull()]
display(analise_funcionario)
```

```python
#Desafio 2
#Depois do tratamento:
# Mostre:
# Média salarial por departamento.
media_salarial=df.groupby("Departamento")["Salario"].mean()
display(media_salarial)
```

```python
#Desafio 3
# Crie um relatório contendo:
# Quantidade de funcionários
# Quantos valores nulos existiam
# Quantos registros duplicados existiam
# Quantas linhas foram removidas
# Quantas linhas ficaram no DataFrame final
analise_funcionario=df["Nome"].count()
nulos=df.isnull().sum().sum()
duplicados=df.drop_duplicates()
cont_linha=len(df) - len(duplicados)
linhas_finais=len(duplicados)

relatorio={
 "Quantidade de funcionários:":analise_funcionario,
 "Quantos valores nulos existiam:":nulos,
 "Quantos registros duplicados existiam:":duplicados,
 "Quantas linhas foram removidas:":cont_linha,
 "Quantas linhas ficaram no DataFrame final":cont_linha
}
display(relatorio)
```
