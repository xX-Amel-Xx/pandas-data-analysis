```python
import pandas as pd
funcionarios=pd.read_csv("funcionarios.csv")
pedidos=pd.read_csv("pedidos.csv")
display(funcionarios,pedidos)
```

```python
#1. INNER JOIN
#Faça um merge() entre df_funcionarios e df_pedidos usando a coluna ID.
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="inner"
    
)
display(df_resultado)
```

```python
#2. LEFT JOIN
#Faça o mesmo, mas usando:
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="left")
display(df_resultado)
```

```python
#2. 3. RIGHT JOIN
#Agora use:
#how="right"
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="right")
display(df_resultado)
```

```python
#4. OUTER JOIN
#Faça:
#how="outer"
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="outer")
display(df_resultado)
```

```python
#5. merge() + seleção
#Faça um merge() inner e mostre somente:
#Nome
#Produto
#Valor
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="inner")
df_resultado=df_resultado[["Nome","Produto","Valor"]]
display(df_resultado)
```

```python
#6. merge() + sort_values()
#Faça um merge() inner e mostre os pedidos ordenados pelo maior valor.
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="inner")
df_resultado=df_resultado.sort_values(by="ID_Pedido",ascending=False)
print(df_resultado)
```

```python
#Exercícios de concat()
#7. Separe os pedidos
#Divida df_pedidos em duas partes:
#Depois junte novamente usando concat().

parte1 = pedidos.iloc[:4]
parte2 = pedidos.iloc[4:]

df_resultado=pd.concat([parte1,parte2])
display(df_resultado)
```

```python
#8. concat() com índice
#Faça novamente usando:
#ignore_index=True
#Observe a diferença no índice.
df_resultado=pd.concat([parte1,parte2],ignore_index=True)
display(df_resultado)
```

```python
#Exercícios de join()
#Primeiro prepare os DataFrames:
#9. join()
#Use join() para juntar funcionarios e pedidos.
#Observe o que acontece quando um funcionário possui mais de um pedido.
analise_f = funcionarios[["ID", "Nome"]].set_index("ID")
analise_pe = pedidos[["ID_Funcionario", "Valor"]].set_index("ID_Funcionario")

df_resultado=analise_f.join(analise_pe)
display(df_resultado)
```

```python
#10. Desafio 
#Faça um merge() left e descubra:
#Quanto cada funcionário gastou no total.
#O resultado deve ter:
#Nome
#Total gasto

df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                    how="left")
df_resultado=df_resultado.groupby(["ID","Nome"]).agg(Total_gasto=("Valor","sum")).reset_index()
display(df_resultado)
```

```python
#No resultado outer, descubra quais registros vieram:
#somente de df_funcionarios
#somente de df_pedidos
#dos dois DataFrames

df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="outer",
                      indicator=True)
display(df_resultado)
```

```python
#Exercícios com seleção
#9.Faça um left merge e mostre somente:
#Nome
#Departamento
#Produto
#Valor

df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="left")
df_resultado=df_resultado[["Nome","Departamento","Produto","Valor"]]
display(df_resultado)
```

```python
#Faça um inner merge e ordene os resultados pelo maior Valor
df_resultado=pd.merge(funcionarios,
                      pedidos,
                    left_on="ID",
                    right_on="ID_Funcionario",
                    how="inner")
df_resultado=df_resultado.sort_values(by="Valor",ascending=False)
display(df_resultado)
```

```python
#11.
#Faça um left merge e descubra quantos funcionários não possuem pedidos.
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="left")
df_resultado=df_resultado["ID_Pedido"].isnull().sum()
display(df_resultado)
```

```python
#Faça um left merge entre funcionários e pedidos.
#Depois:
#Agrupe pelo Nome.
#Calcule o total gasto por cada funcionário.
#Os funcionários sem pedidos devem continuar aparecendo.
df_resultado=pd.merge(funcionarios,
                      pedidos,
                      left_on="ID",
                      right_on="ID_Funcionario",
                      how="left")
df_resultado=df_resultado.groupby("Nome").agg(Total_gasto=("Valor","sum"))
display(df_resultado)
```
