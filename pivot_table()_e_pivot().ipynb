```python
#Crie uma tabela mostrando a média de Salario:
#nas linhas → Departamento
#nas colunas → Cidade
import pandas as pd
funcionarios=pd.read_csv("funcionarios.csv")
pedidos=pd.read_csv("pedidos.csv")
display(funcionarios,pedidos)
```

```python
#1. Crie uma tabela mostrando a média de Salario:
#nas linhas → Departamento
#nas colunas → Cidade
resultado=funcionarios.pivot_table(
index="Departamento",
columns="Cidade",
values="Salario",
aggfunc="mean",
)
display(resultado)
```

```python
#2. Total de salários por departamento e cidade
#Agora faça uma tabela mostrando o total de salários para cada combinação de:
#Departamento
#Cidade
#Você vai precisar mudar a função de agregação.
resultado=funcionarios.pivot_table(
    index="Departamento",
    columns="Cidade",
    values="Salario",
    aggfunc="sum"
)
display(resultado)
```

```python
#3. Crie uma tabela mostrando quantos funcionários existem em cada combinação de:
#Departamento
#Cidade
resultado=funcionarios.pivot_table(
    index="Departamento",
    columns="Cidade",
    values="Nome",
    aggfunc="count"
)
display(resultado)
```

```python
#4. Maior salário
#Crie uma tabela mostrando o maior salário de cada combinação de:
#Departamento
#Cidade
resultado=funcionarios.pivot_table(
    index="Departamento",
    columns="Cidade",
    values="Salario",
    aggfunc="max"
)
display(resultado)
```

```python
#5. Menor salário
#Faça a mesma coisa do exercício anterior, mas mostrando o menor salário.
resultado=funcionarios.pivot_table(
    index="Departamento",
    columns="Cidade",
    values="Salario",
    aggfunc="min"
)
display(resultado)
```

```python
#6.
#Refaça o exercício 1, mas substitua os NaN por 0.
#Observe como o resultado muda.
resultado=funcionarios.pivot_table(
index="Departamento",
columns="Cidade",
values="Salario",
aggfunc="mean",
fill_value=0
)
display(resultado)
```

```python
#7. Crie uma pivot_table() mostrando a média salarial por cidade e departamento, mas agora inverta:
#nas linhas → Cidade
#nas colunas → Departamento
resultado=funcionarios.pivot_table(
index="Cidade",
columns="Departamento",
values="Salario",
aggfunc="mean"
)
display(resultado)
```

```python
#8.
#Crie um DataFrame pequeno contendo:
#Nome
#Departamento
#Cidade
#Salario
#com funcionários suficientes para que cada
#combinação de Departamento + Cidade apareça uma única vez.

dados = pd.DataFrame({
    "Nome": ["Ana", "João", "Maria", "Pedro"],
    "Departamento": ["TI", "TI", "RH", "RH"],
    "Cidade": ["Brasília", "Goiânia", "Brasília", "Goiânia"],
    "Salario": [5000, 4500, 3500, 4000]
})
display(dados)

resultado=dados.pivot(
    index="Departamento",
    columns="Cidade",
    values="Salario"
)
display(resultado)
```
